```python
import numpy as np
import matplotlib.pyplot as plt
from function_library import * 
```

```python
directory = "Transverse_Momentum_Plots"

ylim = 0.25
plt.rcParams["figure.figsize"] = (5,5)
xlabel = "$p_{T}$ [GeV/c]"
ylabel = "$\sigma_{p_{T}} / p_T$"
```

#### Proton:
- transverse momentum resolution
- for different momenta ( 0 - 4 GeV/c)
- for different polar angles (90°, 67.5°, 45°)
- 4 hits in the outer layers of the ITS2

```python
proton_momenta = np.arange(0.2, 4, 0.1)                     # momentum of particle in GeV/c
proton_mass = 1                                             # mass of particle in GeV/c^2 

N = 4                                                       # Number of layers with hits
theta_list = [90, 67.5, 45]                                 # polar angles of the particle

proton_pT_4hits = np.array([[transverse_momentum_reso(momentum, proton_mass, N, theta) for momentum in proton_momenta] for theta in theta_list])
proton_pT_4hits.shape
```

```python
plt.plot(proton_momenta, proton_pT_4hits[0], ".")
plt.xlabel(xlabel)
plt.ylabel(ylabel)
plt.ylim(0, ylim)
plt.text(2.5, 0.15, "Proton with 4 hits")
plt.text(2.5, 0.14, fr"$\theta = {theta_list[0]}$°")
plt.tick_params(top=True, right=True, direction="in")
plt.savefig(f"{directory}/Proton_pT.png", dpi = 200)
```

```python
for i, theta in enumerate(theta_list):
    plt.plot(proton_momenta, proton_pT_4hits[i], ".", label=f"polarangle = {theta} °")
plt.legend()
plt.xlabel(xlabel)
plt.ylabel(ylabel)
plt.ylim(0, ylim)
plt.text(2.5, 0.15, "Proton with 4 hits")
plt.tick_params(top=True, right=True, direction="in")
plt.savefig(f"{directory}/Proton_pT_angle.png", dpi = 200)
```
