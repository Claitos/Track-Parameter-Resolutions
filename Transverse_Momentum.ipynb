```python
import numpy as np
import matplotlib.pyplot as plt
from function_library import * 
```

```python
directory = "Transverse_Momentum_Plots"
ylim = 0.25
```

#### Proton:
- transverse momentum resolution
- for different momenta ( 0 - 4 GeV/c)
- for different polar angles (90°, 67.5°, 45°)
- 4 hits in the outer layers of the ITS2

```python
proton_momenta = np.arange(0.2, 4, 0.1)                   # momentum of particle in GeV/c
proton_mass = 1                                             # mass of particle in GeV/c^2 

N = 4                                                       # Number of layers with hits
theta_list = [90, 67.5, 45]                                 # polar angles of the particle

proton_pT_det_list = []
proton_pT_MS_list = []
for momentum in proton_momenta:
    pT_det, pT_MS = transverse_momentum_reso(momentum, proton_mass, N)
    proton_pT_det_list.append(pT_det); proton_pT_MS_list.append(pT_MS)

proton_pT_det_list = np.array(proton_pT_det_list)
proton_pT_MS_list = np.array(proton_pT_MS_list)
```

```python
plt.plot(proton_momenta, total_transverse_resolution(proton_pT_det_list, proton_pT_MS_list, theta_list[0]), ".")
plt.xlabel("$p_{T}$ [GeV/c]")
plt.ylabel("$\sigma_{p_{T}} / p_T$")
plt.ylim(0, ylim)
plt.text(2.5, 0.15, "Proton with 4 hits")
plt.text(2.5, 0.14, fr"$\theta = 90$°")
plt.tick_params(top=True, right=True, direction="in")
plt.savefig(f"{directory}/Proton_pT.png", dpi = 200)
```

```python
for theta in theta_list:
    proton_pT_list = total_transverse_resolution(proton_pT_det_list, proton_pT_MS_list, theta)
    plt.plot(proton_momenta, proton_pT_list, ".", label=f"polarangle = {theta} °")
plt.legend()
plt.xlabel("$p_{T}$ [GeV/c]")
plt.ylabel("$\sigma_{p_{T}} / p_T$")
plt.ylim(0, ylim)
plt.text(2.5, 0.15, "Proton with 4 hits")
plt.tick_params(top=True, right=True, direction="in")
plt.savefig(f"{directory}/Proton_pT_angle.png", dpi = 200)
```
