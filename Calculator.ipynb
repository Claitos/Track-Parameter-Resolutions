```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def theta(eta):
    return 2*np.arctan(np.exp(-eta))

def rad_to_deg(rad):
    return rad * 180 / np.pi

def deg_to_rad(deg):
    return deg * np.pi / 180

print(theta(0.9))
print(rad_to_deg(theta(0.9)))
```

```python
momentum_reso_off = 1 / np.sin(theta(0.9))**(1/2)
pos_reso_para_off = 1 / np.sin(theta(0.9))**(1/2)
pos_reso_straight_off = 1 / np.sin(theta(0.9))**(3/2)
print(pos_reso_para_off, pos_reso_straight_off)
```

```python
real_mean_angle = (0.5*np.pi - theta(0.9)) / 2  + theta(0.9)
print(real_mean_angle)
print(rad_to_deg(real_mean_angle))

momentum_reso_off = 1 / np.sin(real_mean_angle)**(1/2)
pos_reso_para_off = 1 / np.sin(real_mean_angle)**(1/2)
pos_reso_straight_off = 1 / np.sin(real_mean_angle)**(3/2)
print(pos_reso_para_off, pos_reso_straight_off)
```
